```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        return 0 if b == 0 else 1
    delta = b ** 2 - 4 * a * c
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0
```

```python
def point_in_quadrilateral(quad, p):
    def cross(o, a, b):
        return (a[0]-o[0])*(b[1]-o[1]) - (a[1]-o[1])*(b[0]-o[0])
    n = len(quad)
    signs = []
    for i in range(n):
        a = quad[i]
        b = quad[(i+1) % n]
        c = cross(a, b, p)
        signs.append(c)
    return all(s >= 0 for s in signs) or all(s <= 0 for s in signs)
```

```python
import math
def tinhgoc(x1, y1, x2, y2, x3, y3):
    ba = (x1-x2, y1-y2)
    bc = (x3-x2, y3-y2)
    dot = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    cos_angle = dot / (mag_ba * mag_bc)
    cos_angle = max(-1, min(1, cos_angle))
    return math.degrees(math.acos(cos_angle))
```

```python
def C(n, k):
    if k < 0 or k > n:
        return 0
    if k == 0 or k == n:
        return 1
    k = min(k, n-k)
    result = 1
    for i in range(k):
        result = result * (n-i) // (i+1)
    return result
```

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        open_fare = 12000
        km_fare = 8500
    else:
        open_fare = 15000
        km_fare = 11000

    fare = open_fare + km * km_fare + wait_minutes * 500

    if start_hour >= 22 or start_hour < 6:
        fare *= 1.2

    if km > 30:
        fare *= 0.9

    return round(fare)
```

```python
def has_all_distinct_pairs(x):
    s = str(abs(x))
    if len(s) % 2 != 0:
        return False
    pairs = [s[i:i+2] for i in range(0, len(s), 2)]
    return len(pairs) == len(set(pairs))

def sum_1_to_n(n):
    if n <= 0:
        return 0
    return sum(x for x in range(1, n+1) if has_all_distinct_pairs(x))
```
