```python
def mangdep(lst):
    count = {}
    for x in lst:
        count[x]= count.get(x,0)+1
    y=0
    for x,f in count.items:
        if f >= x:
            y += (f - x)
        else:
            y += f 
    return y
```

```python
def sapxep(lst):
    chan=[n for n in lst if n%2==0]
    le=[n for n in lst if n%2!=0]
    chan.sort()
    le.sort(reverse=True)
    newlst=chan + le
    return newlst
        
```

```python
def tinhtong(lst,k):
    tong = 0
    for i in lst:
        if len(str(i))==k:
            tong+= i
    return tong
lst=list(map(int,input().split()))
k = int(input())
print(tinhtong(lst,k))
```

```python
def xoatu(xau):
    cactu = xau.split()
    if not cactu:
        return ''
    ketqua = cactu[0]
    for i in range(1,len(cactu)):
        if cactu[i]!= cactu[i-1]:
            ketqua.append(cactu[i])
    return ''.join(ketqua)
```

```python
import re

def tinh_toan_so_trong_van_ban(van_ban):
    # Tìm tất cả các con số (đang ở dạng chuỗi)
    cac_so_str = re.findall(r'\d+', van_ban)
    
    # Ép kiểu tất cả sang số nguyên (int)
    cac_so = [int(x) for x in cac_so_str]
    
    if not cac_so:
        return "Không tìm thấy số nào", 0, 0
        
    tong = sum(cac_so)
    trung_binh = tong / len(cac_so)
    
    return cac_so, tong, trung_binh
```
