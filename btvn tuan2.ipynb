## b5

```python
import math
a, b, c = map(float, input().split())
if a == 0:
    print("giải như phương trình bậc 1 ở câu 4")
else:
    delta = b**2 - 4 * a * c
    if delta < 0:
        print("vô nghiệm")
    elif delta == 0:
        x = -b / (2*a)
        print("nghiệm kép", x)
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print("có 2 nghiệm ", x1, x2)
    
```

## b8

```python
a, b , c , d , m , n = map(float, input().split())
D = a * d - b * c
Dx = m * d - b * n 
Dy = a * n - m * c
if D == 0:
    if Dx == 0 and Dy == 0:
        print("hệ vô số nghiệm")
    else:
        print("vô nghiệm")
else:
    x = Dx / D
    y = Dy / D
    print("x =", x, "y =", y)
```

## b10

```python
import math
xi, yi = map(float, input("Nhập tọa độ tâm x y: ").split())
r = float(input("Nhập bán kính R: "))
xa, ya = map(float, input("Nhập tọa độ điểm A x y: ").split())
d = math.sqrt((xa - xi)**2 + (ya - yi)**2)
if d < r:
    print("Điểm A nằm TRONG đường tròn.")
elif d == r: # (*) Lưu ý nhỏ về sai số
    print("Điểm A nằm TRÊN (thuộc) đường tròn.")
else:
    print("Điểm A nằm NGOÀI đường tròn.")
```
