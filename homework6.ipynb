```python
# Bước 1: Nhập kích thước n của ma trận vuông
n = int(input())
# Bước 2: Nhập ma trận n x n
matran = []
for i in range(n):
    hang = list(map(int,input().split()))
    matran.append(hang)
    # Bước 3: Tìm đường chéo chính
# Đặc điểm: Chỉ số hàng bằng chỉ số cột (i == j)
dcc = []
for i in range(n):
    dcc.append(matran[i][i])
# Bước 4: Tìm đường chéo phụ
# Đặc điểm: Tổng chỉ số hàng và cột luôn bằng n - 1 (j = n - 1 - i)
dcp=[]
for i in range(n):
    dcp.append(matran[i][n-1-i])
print(*dcc)
print(*dcp)

```

```python
def vitri(lst,k:int):
    for n in lst:
        if n == k:
            return lst.index(n)
    else:
        return -1
lst = list(map(int,input().split()))
k = int(input())
print(vitri(lst,k))
```

```python
def anh_xa_vi_tri():
    
    lst = input().split()
    mapping = {}

   
    for i, x in enumerate(lst):
        if x not in mapping:
            mapping[x] = [] 
        mapping[x].append(i)

    
    for x in mapping:
        mapping[x] = tuple(mapping[x])
    
    return mapping


print(anh_xa_vi_tri())
```
